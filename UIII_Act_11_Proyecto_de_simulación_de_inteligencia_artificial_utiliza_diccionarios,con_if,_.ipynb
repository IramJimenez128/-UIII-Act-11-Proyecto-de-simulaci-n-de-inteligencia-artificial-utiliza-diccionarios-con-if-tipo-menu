### Diccionario Conductor
### Lista conductor


```python
# --- SISTEMA DE IA: CONTROL DE TRANSPORTES V2.0 (MODULAR) ---

# Inicialización de las estructuras de datos solicitadas
# 'conductor' es la lista que guardará todos los registros
conductor = []
# 'los_conductores' es el diccionario que inicia vacío para capturar entradas
los_conductores = {}

def agregar_conductor():
    """Captura los 6 atributos y los guarda en el diccionario 'los_conductores'."""
    print("\n--- Registro de Nuevo Conductor ---")

    # Entrada de datos interactiva [5]
    licencia = input("Tipo de licencia (A/B/C): ")
    ruta = input("Ruta asignada: ")
    placas = input("Placas del vehículo: ")
    horas = int(input("Horas de manejo acumuladas: "))
    empresa = input("Empresa de transporte: ")
    record = input("Récord de seguridad (limpio/riesgo): ").lower()

    # Se crea el diccionario con los nombres de clave (keys) [1]
    los_conductores = {
        "licencia": licencia,
        "ruta": ruta,
        "placas": placas,
        "horas": horas,
        "empresa": empresa,
        "record": record
    }

    # Se añade el diccionario a la lista global 'conductor' [6]
    conductor.append(los_conductores)
    print("IA: Registro completado exitosamente.")

def listar_conductores():
    """Presenta la información de la lista 'conductor' en forma de tabla [7]."""
    if not conductor:
        print("\nIA: La lista de conductores está vacía.")
    else:
        # Formato de tabla utilizando f-strings para alineación [7]
        encabezado = f"{'ID':<4} | {'Empresa':<15} | {'Placas':<10} | {'Horas':<8} | {'Récord':<12}"
        print("\n" + "=" * len(encabezado))
        print(encabezado)
        print("-" * len(encabezado))

        # Recorrido de la lista con ciclo for [8]
        indice = 1
        for c in conductor:
            # Acceso a los valores mediante las claves del diccionario [9]
            fila = f"{indice:<4} | {c['empresa']:<15} | {c['placas']:<10} | {c['horas']:<8} | {c['record']:<12}"
            print(fila)
            indice += 1
        print("=" * len(encabezado))

def analisis_ia():
    """Simulación de IA que analiza riesgos basados en los datos almacenados."""
    if not conductor:
        print("\nIA: Sin datos para analizar.")
    else:
        print("\n--- INFORME DE RIESGOS (IA) ---")
        for c in conductor:
            print(f">> Analizando unidad [{c['placas']}]...")

            # Lógica condicional para toma de decisiones [10]
            if c['horas'] > 12:
                print("   ESTADO: Alerta de fatiga. Conductor debe detenerse.")
            elif c['record'] == "riesgo":
                print("   ESTADO: Antecedentes críticos. Requiere inspección.")
            else:
                print("   ESTADO: Operación normal y segura.")

# --- MENÚ PRINCIPAL (Ciclo While True) [11] ---
while True:
    print("\n******************************************")
    print("      MENÚ IA: ASISTENTE DE TRANSPORTE")
    print("******************************************")
    print("1. Agregar Conductor")
    print("2. Ver Tabla de 'conductor'")
    print("3. Ejecutar Análisis IA")
    print("4. Salir")

    opcion = input("\nSeleccione una opción: ")

    if opcion == "1":
        agregar_conductor()
    elif opcion == "2":
        listar_conductores()
    elif opcion == "3":
        analisis_ia()
    elif opcion == "4":
        print("IA: Desconectando sensores... ¡Buen viaje!")
        break # Termina la ejecución [12]
    else:
        print("IA: Opción inválida.")
print("Iram Jimenez 2H 0079")
```
