```python
# --- SISTEMA DE IA: ASISTENTE DE CONTROL DE CONDUCTORES V1.0 ---

# Inicialización de las estructuras de datos
# conductores: Lista global que almacenará los registros
conductores = []
# datos_conductor: Diccionario que inicia vacío para capturar entradas
datos_conductor = {}

def agregar_conductor():
    """Captura los 6 atributos requeridos y los guarda en un diccionario."""
    print("\n--- Registro de Nuevo Conductor ---")

    # Entrada de datos (Key: Value)
    licencia = input("Tipo de licencia (A/B/C): ")
    ruta = input("Ruta asignada: ")
    placas = input("Placas del vehículo: ")
    horas = int(input("Horas de manejo acumuladas: "))
    empresa = input("Empresa de transporte: ")
    record = input("Récord de seguridad (limpio/advertencia/critico): ").lower()

    # Creación del diccionario con los 6 atributos [1, 2]
    datos_conductor = {
        "licencia": licencia,
        "ruta": ruta,
        "placas": placas,
        "horas": horas,
        "empresa": empresa,
        "record": record
    }

    # Se añade el diccionario a la lista global [3, 4]
    conductores.append(datos_conductor)
    print("IA: Registro completado y almacenado en la base de datos.")

def listar_conductores():
    """Muestra los conductores en un formato de tabla bien distribuido."""
    if not conductores:
        print("\nIA: La lista de conductores está vacía actualmente.")
    else:
        # Definición de anchos de columna para la tabla
        encabezado = f"{'ID':<4} | {'Empresa':<15} | {'Placas':<10} | {'Licencia':<10} | {'Horas':<8} | {'Récord':<12}"
        print("\n" + "=" * len(encabezado))
        print(encabezado)
        print("-" * len(encabezado))

        # Ciclo for para recorrer la lista de diccionarios [5, 6]
        indice = 1
        for c in conductores:
            fila = f"{indice:<4} | {c['empresa']:<15} | {c['placas']:<10} | {c['licencia']:<10} | {c['horas']:<8} | {c['record']:<12}"
            print(fila)
            indice += 1
        print("=" * len(encabezado))

def analizar_conductores_ia():
    """Simulación de IA que analiza riesgos basados en horas y récord."""
    if not conductores:
        print("\nIA: No hay datos suficientes para realizar un análisis de seguridad.")
    else:
        print("\n--- MÓDULO DE ANÁLISIS LOGÍSTICO (IA) ---")
        for c in conductores:
            print(f"\n>> Analizando vehículo [{c['placas']}] de la empresa {c['empresa']}:")

            # Lógica de IA utilizando estructuras selectivas (if, elif, else) [7, 8]
            if c['horas'] > 12:
                print("   IA ALERTA: Fatiga detectada. El conductor debe tomar un descanso obligatorio.")
            elif c['record'] == "critico":
                print("   IA ALERTA: Récord peligroso. Se recomienda suspensión temporal de la ruta.")
            elif c['record'] == "limpio" and c['horas'] <= 8:
                print("   IA DIAGNÓSTICO: Desempeño óptimo. Candidato para rutas de larga distancia.")
            else:
                print("   IA DIAGNÓSTICO: Estado estándar. Mantener monitoreo preventivo.")

# --- CICLO PRINCIPAL (MENÚ INTERACTIVO) ---
while True:
    print("\n******************************************")
    print("   MENÚ DE GESTIÓN DE CONDUCTORES (IA)")
    print("******************************************")
    print("1. Agregar Conductor")
    print("2. Listar Conductores (Tabla)")
    print("3. Ejecutar Análisis de Seguridad IA")
    print("4. Salir")

    opcion = input("\nSeleccione una opción (1-4): ")

    if opcion == "1":
        agregar_conductor()
    elif opcion == "2":
        listar_conductores()
    elif opcion == "3":
        analizar_conductores_ia()
    elif opcion == "4":
        print("IA: Cerrando sistemas de control. ¡Buen viaje!")
        break # Rompe el ciclo para finalizar [9]
    else:
        print("IA: Error. Opción no válida, intente de nuevo.")

print("\n--- SESIÓN FINALIZADA ---")
print("Programa realizado ´por Iram jimenez 0079")
```
